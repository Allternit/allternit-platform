# S1 typed-decisions fine-tune (Kaggle GPU variant)

Same recipe as `run-finetune.sh`. Upload the `system-one export` directory as a Kaggle dataset named `s1-export` and this repo's `finetune_laya.py` as `s1-finetune`. Download `/kaggle/working/ckpt` afterwards, run `system-one calibrate --q26` on `scored-tune.jsonl` / `scored-cert.jsonl` locally, and swap it in only if a bank passes.

The export holds raw user text (opt-in only). Keep the Kaggle dataset private and delete it after the run.
